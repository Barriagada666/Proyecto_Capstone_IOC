```python
#!/usr/bin/env python3
"""
Generador de Gráfico Burndown para Sprint 1 - Versión Google Colab
Genera un gráfico de burndown basado en los datos del Sprint 1 Backlog
"""

# Importar librerías necesarias
import matplotlib.pyplot as plt
import pandas as pd
from datetime import datetime

# Configurar matplotlib para Colab
%matplotlib inline

# Datos del Sprint 1
sprint_data = {
    'Semana': ['Inicio', 'Semana 1', 'Semana 2', 'Semana 3', 'Semana 4'],
    'SP_Planeados': [0, 10, 10, 10, 11],  # Story Points planeados por semana
    'SP_Reales': [0, 12, 10, 10, 9],      # Story Points realmente completados
    'SP_Restantes': [41, 29, 19, 9, 0],   # Story Points restantes
    'Burndown_Ideal': [41, 30.75, 20.5, 10.25, 0]  # Burndown ideal (línea recta)
}

# Crear DataFrame
df = pd.DataFrame(sprint_data)

# Configurar el estilo del gráfico
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(14, 8))

# Graficar la línea de burndown ideal
ax.plot(df['Semana'], df['Burndown_Ideal'],
        marker='o',
        linestyle='--',
        color='#28a745',  # Verde
        linewidth=2.5,
        markersize=10,
        label='Burndown Ideal',
        alpha=0.7)

# Graficar la línea de esfuerzo restante (burndown real)
ax.plot(df['Semana'], df['SP_Restantes'],
        marker='s',
        linestyle='-',
        color='#007bff',  # Azul
        linewidth=3,
        markersize=10,
        label='Esfuerzo Restante (SP)',
        alpha=0.9)

# Añadir valores en los puntos
for i, (semana, restante, ideal) in enumerate(zip(df['Semana'], df['SP_Restantes'], df['Burndown_Ideal'])):
    ax.text(i, restante + 1.5, f'{restante} SP',
            ha='center', va='bottom', fontsize=10, color='#007bff', fontweight='bold')
    ax.text(i, ideal - 1.5, f'{ideal:.1f} SP',
            ha='center', va='top', fontsize=10, color='#28a745', alpha=0.7, fontweight='bold')

# Configurar etiquetas y título
ax.set_xlabel('Semana del Sprint', fontsize=13, fontweight='bold')
ax.set_ylabel('Story Points Restantes', fontsize=13, fontweight='bold')
ax.set_title('Sprint 1 - Gráfico de Burndown\n(8 Sept - 3 Oct 2025, Total: 41 SP)',
             fontsize=16, fontweight='bold', pad=20)

# Configurar el eje Y
ax.set_ylim(-2, 45)
ax.set_yticks(range(0, 45, 5))

# Añadir grid
ax.grid(True, alpha=0.4, linestyle=':', linewidth=1)

# Añadir leyenda
ax.legend(loc='upper right', fontsize=11, framealpha=0.95, shadow=True)

# Añadir información adicional
info_text = (
    f"Período: 8 Sept - 3 Oct 2025 (4 semanas)\n"
    f"Días hábiles: 18 días\n"
    f"Total Story Points: 41 SP\n"
    f"Historias comprometidas: 5"
)
ax.text(0.02, 0.98, info_text,
        transform=ax.transAxes,
        fontsize=10,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.6, edgecolor='black', linewidth=1.5))

# Ajustar layout
plt.tight_layout()

# Guardar el gráfico
output_file = 'sprint1_grafico_burndown.png'
plt.savefig(output_file, dpi=300, bbox_inches='tight', facecolor='white')
print(f"✅ Gráfico guardado en: {output_file}")

# Mostrar el gráfico
plt.show()

# Generar reporte en consola
print("\n" + "="*70)
print("REPORTE DEL SPRINT 1 - GRÁFICO DE BURNDOWN".center(70))
print("="*70)
print(f"\n📅 Período: 8 Septiembre - 3 Octubre 2025")
print(f"⏱️  Duración: 4 semanas (18 días hábiles)")
print(f"📊 Total Story Points: 41 SP\n")
print(df.to_string(index=False))
print("\n" + "="*70)
print("\n✅ Historias completadas:")
print("  • IOC-021: Iniciar sesión (5 SP)")
print("  • IOC-022: Cerrar sesión (2 SP)")
print("  • IOC-023: Navegación entre secciones (8 SP)")
print("  • IOC-001: Cargar y validar archivos (13 SP)")
print("  • IOC-006: Visualizar dashboard con KPIs (13 SP)")
print("="*70)

# OPCIONAL: Descargar el archivo automáticamente
print("\n💾 Para descargar el gráfico, ejecuta la siguiente celda:")
```

```python
# Análisis adicional del progreso
print("\n📈 ANÁLISIS DE VELOCIDAD DEL SPRINT\n")
print("="*70)

# Calcular métricas
total_sp = 41
completed_weeks = 4
velocity_per_week = df['Actual_SP'][1:].mean()
deviation = df['Remaining_SP'] - df['Ideal_Burndown']

metrics_df = pd.DataFrame({
    'Week': df['Week'][1:],
    'Planned SP': df['Planned_SP'][1:],
    'Actual SP': df['Actual_SP'][1:],
    'Remaining': df['Remaining_SP'][1:],
    'Ideal': df['Ideal_Burndown'][1:],
    'Deviation': deviation[1:].round(2)
})

print(metrics_df.to_string(index=False))
print("\n" + "="*70)
print(f"\n📊 Velocidad promedio: {velocity_per_week:.2f} SP/semana")
print(f"🎯 Estado final: {'✅ COMPLETADO' if df['Remaining_SP'].iloc[-1] == 0 else '⚠️ INCOMPLETO'}")
print(f"📉 Desviación final: {deviation.iloc[-1]:.2f} SP")
print("="*70)
```

```python
# Celda adicional para análisis con horas extras
print("\n" + "="*70)
print("⚠️  ANÁLISIS DE VELOCIDAD AJUSTADA (CON HORAS EXTRAS)".center(70))
print("="*70)

# Datos reales
total_sp_completed = 41
planned_weeks = 4
overtime_days = 2  # Días extras trabajados (fin de semana)
business_days_per_week = 4.5  # 18 días ÷ 4 semanas

# Calcular capacidad real
total_planned_days = 18  # Días hábiles normales
total_actual_days = 18 + overtime_days  # Incluyendo fin de semana extra
actual_weeks_equivalent = total_actual_days / business_days_per_week

# Velocidades
nominal_velocity = total_sp_completed / planned_weeks
sustainable_velocity = total_sp_completed / actual_weeks_equivalent

print(f"\n📅 Tiempo planeado: {planned_weeks} semanas ({total_planned_days} días hábiles)")
print(f"⏰ Tiempo real trabajado: {actual_weeks_equivalent:.2f} semanas ({total_actual_days} días)")
print(f"🔴 Horas extras: {overtime_days} días de fin de semana\n")

print(f"📊 Velocidad NOMINAL: {nominal_velocity:.2f} SP/semana")
print(f"   └─ ❌ NO SOSTENIBLE (incluye horas extras)\n")

print(f"✅ Velocidad SOSTENIBLE: {sustainable_velocity:.2f} SP/semana")
print(f"   └─ Esta es tu VERDADERA capacidad\n")

# Impacto en Sprint 2
recommended_sp_sprint2 = sustainable_velocity * planned_weeks
over_commitment = total_sp_completed - recommended_sp_sprint2

print("="*70)
print(f"\n🎯 RECOMENDACIÓN PARA SPRINT 2:")
print(f"   • Comprométete a: {recommended_sp_sprint2:.0f} SP (no {total_sp_completed} SP)")
print(f"   • Sobre-compromiso del Sprint 1: {over_commitment:.0f} SP")
print(f"   • Riesgo de burnout: {'🔴 ALTO' if over_commitment > 5 else '🟡 MEDIO'}")
print("="*70)
```

```python
#!/usr/bin/env python3
"""
Generador de Gráfico Burndown para Sprint 1 - Versión CON HORAS EXTRAS
Refleja el trabajo adicional de fin de semana
"""

# Importar librerías necesarias
import matplotlib.pyplot as plt
import pandas as pd
from datetime import datetime

# Configurar matplotlib para Colab
%matplotlib inline

# Datos REALES del Sprint 1 (incluyendo overtime)
sprint_data_real = {
    'Semana': ['Inicio', 'Semana 1', 'Semana 2', 'Semana 3', 'Semana 4', 'Sem 4+Extra'],
    'SP_Planeados': [0, 10, 10, 10, 11, 0],
    'SP_Reales': [0, 12, 10, 10, 5, 4],  # 5 SP en semana 4 normal, 4 SP en fin de semana
    'SP_Restantes': [41, 29, 19, 9, 4, 0],
    'Burndown_Ideal': [41, 30.75, 20.5, 10.25, 0, 0]
}

# Crear DataFrame
df_real = pd.DataFrame(sprint_data_real)

# Configurar el estilo del gráfico
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(14, 8))

# Línea ideal (solo hasta semana 4)
ax.plot(df_real['Semana'][:5], df_real['Burndown_Ideal'][:5],
        marker='o', linestyle='--', color='#28a745', linewidth=2.5,
        markersize=10, label='Burndown Ideal', alpha=0.7)

# Línea real (hasta semana 4)
ax.plot(df_real['Semana'][:5], df_real['SP_Restantes'][:5],
        marker='s', linestyle='-', color='#007bff', linewidth=2.5,
        markersize=10, label='Esfuerzo Restante (Horario Normal)', alpha=0.9)

# Extensión por horas extras
ax.plot(df_real['Semana'][4:], df_real['SP_Restantes'][4:],
        marker='D', linestyle=':', color='#dc3545', linewidth=3,
        markersize=10, label='⚠️ Horas Extras (Fin de Semana)', alpha=0.9)

# Zona de horas extras (sombreado)
ax.axvspan(4, 5, alpha=0.2, color='red', label='Zona de Overtime')

# Línea vertical en Semana 4
ax.axvline(x=4, color='orange', linestyle='--', linewidth=2, alpha=0.6)
ax.text(4.05, 25, 'Fin del Sprint\n(Planeado)', fontsize=9, color='orange',
        fontweight='bold', va='center')

# Añadir valores en los puntos
for i, (semana, restante) in enumerate(zip(df_real['Semana'], df_real['SP_Restantes'])):
    color = '#dc3545' if i == 5 else '#007bff'
    ax.text(i, restante + 1.5, f'{restante} SP',
            ha='center', va='bottom', fontsize=10, color=color, fontweight='bold')

# Configurar etiquetas y título
ax.set_xlabel('Tiempo del Sprint', fontsize=13, fontweight='bold')
ax.set_ylabel('Story Points Restantes', fontsize=13, fontweight='bold')
ax.set_title('Sprint 1 - Gráfico de Burndown (CON HORAS EXTRAS)\n⚠️ Se trabajó fin de semana adicional',
             fontsize=16, fontweight='bold', pad=20, color='darkred')

# Configurar el eje Y
ax.set_ylim(-2, 45)
ax.set_yticks(range(0, 45, 5))

# Añadir grid
ax.grid(True, alpha=0.4, linestyle=':', linewidth=1)

# Advertencia
warning_text = (
    f"⚠️ ALERTA DE SOSTENIBILIDAD\n"
    f"Horas extras: 2 días (fin de semana)\n"
    f"Velocidad aparente: 10.25 SP/sem\n"
    f"Velocidad real: ~9.2 SP/sem\n"
    f"Estado: NO SOSTENIBLE"
)
ax.text(0.02, 0.98, warning_text,
        transform=ax.transAxes, fontsize=10,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='salmon', alpha=0.7,
                  edgecolor='darkred', linewidth=2))

# Añadir leyenda
ax.legend(loc='upper right', fontsize=10, framealpha=0.95, shadow=True)

# Ajustar layout
plt.tight_layout()

# Guardar el gráfico
output_file = 'sprint1_burndown_REAL.png'
plt.savefig(output_file, dpi=300, bbox_inches='tight', facecolor='white')
print(f"✅ Gráfico guardado en: {output_file}")

# Mostrar el gráfico
plt.show()

# Reporte en consola
print("\n" + "="*70)
print("⚠️ REPORTE REAL DEL SPRINT 1 - CON HORAS EXTRAS".center(70))
print("="*70)
print(f"\n📅 Período planeado: 8 Septiembre - 3 Octubre 2025 (4 semanas)")
print(f"📅 Período real: 8 Septiembre - 5 Octubre 2025 (4 semanas + 2 días)")
print(f"⏱️  Días hábiles planeados: 18 días")
print(f"⏱️  Días realmente trabajados: 20 días")
print(f"🔴 Horas extras: 2 días de fin de semana\n")
print(df_real.to_string(index=False))
print("\n" + "="*70)
print("\n📊 ANÁLISIS DE VELOCIDAD:")
print(f"   • Velocidad nominal: 10.25 SP/semana ❌ (incluye overtime)")
print(f"   • Velocidad sostenible: 9.2 SP/semana ✅ (sin overtime)")
print(f"   • Sobre-compromiso: 4 SP")
print("\n🎯 RECOMENDACIÓN PARA SPRINT 2:")
print(f"   • Comprometerse a: 33-37 SP (NO 41 SP)")
print(f"   • Incluir buffer de seguridad: 10%")
print(f"   • Evitar trabajo en fin de semana")
print("="*70)
```
