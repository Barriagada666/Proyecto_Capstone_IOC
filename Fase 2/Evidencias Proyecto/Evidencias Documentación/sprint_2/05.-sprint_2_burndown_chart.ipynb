```python
#!/usr/bin/env python3
"""
Generador de Gráfico Burndown para Sprint 2 - Versión Google Colab
Genera un gráfico de burndown basado en los datos del Sprint 2 Backlog
Sprint 2: Interactividad Analítica (6 Oct - 24 Oct 2025)
"""

# Importar librerías necesarias
import matplotlib.pyplot as plt
import pandas as pd
from datetime import datetime

# Configurar matplotlib para Colab
# %matplotlib inline  # Descomentar si se ejecuta en Google Colab

# Datos del Sprint 2
# Progreso basado en tareas completadas hasta el 13 de octubre
sprint_data = {
    'Semana': ['Inicio', 'Semana 1\n(6-10 Oct)', 'Semana 2\n(13-17 Oct)', 'Semana 3\n(20-24 Oct)'],
    'SP_Planeados': [0, 7, 7, 7],  # Story Points planeados por semana (~7 SP/semana)
    'SP_Reales': [0, 5, 0, 0],      # Story Points realmente completados (hasta 13-Oct)
    'SP_Restantes': [21, 16, 16, 16],   # Story Points restantes (actualizado al 13-Oct)
    'Burndown_Ideal': [21, 14, 7, 0]  # Burndown ideal (línea recta)
}

# Crear DataFrame
df = pd.DataFrame(sprint_data)

# Configurar el estilo del gráfico
plt.style.use('seaborn-v0_8-whitegrid')
fig, ax = plt.subplots(figsize=(14, 8))

# Graficar la línea de burndown ideal
ax.plot(df['Semana'], df['Burndown_Ideal'],
        marker='o',
        linestyle='--',
        color='#28a745',  # Verde
        linewidth=2.5,
        markersize=10,
        label='Burndown Ideal',
        alpha=0.7)

# Graficar la línea de esfuerzo restante (burndown real)
ax.plot(df['Semana'], df['SP_Restantes'],
        marker='s',
        linestyle='-',
        color='#007bff',  # Azul
        linewidth=3,
        markersize=10,
        label='Esfuerzo Restante (SP)',
        alpha=0.9)

# Añadir valores en los puntos
for i, (semana, restante, ideal) in enumerate(zip(df['Semana'], df['SP_Restantes'], df['Burndown_Ideal'])):
    ax.text(i, restante + 0.8, f'{restante} SP',
            ha='center', va='bottom', fontsize=10, color='#007bff', fontweight='bold')
    ax.text(i, ideal - 0.8, f'{ideal:.1f} SP',
            ha='center', va='top', fontsize=10, color='#28a745', alpha=0.7, fontweight='bold')

# Añadir línea de proyección (si continúa el ritmo actual)
# Proyección pesimista si no se avanza en Semana 2
proyeccion = [21, 16, 16, 16]
ax.plot(df['Semana'], proyeccion,
        marker='x',
        linestyle=':',
        color='#dc3545',  # Rojo
        linewidth=2,
        markersize=8,
        label='Proyección Pesimista',
        alpha=0.6)

# Configurar etiquetas y título
ax.set_xlabel('Semana del Sprint', fontsize=13, fontweight='bold')
ax.set_ylabel('Story Points Restantes', fontsize=13, fontweight='bold')
ax.set_title('Sprint 2 - Gráfico de Burndown (Interactividad Analítica)\n(6 Oct - 24 Oct 2025, Total: 21 SP)',
             fontsize=16, fontweight='bold', pad=20)

# Configurar el eje Y
ax.set_ylim(-1, 24)
ax.set_yticks(range(0, 25, 3))

# Añadir grid
ax.grid(True, alpha=0.4, linestyle=':', linewidth=1)

# Añadir leyenda
ax.legend(loc='upper right', fontsize=11, framealpha=0.95, shadow=True)

# Añadir información adicional
info_text = (
    f"Período: 6 Oct - 24 Oct 2025 (3 semanas)\n"
    f"Días hábiles: 15 días\n"
    f"Total Story Points: 21 SP\n"
    f"Historias comprometidas: 2\n"
    f"Estado actual: 14 Oct (Semana 2)"
)
ax.text(0.02, 0.98, info_text,
        transform=ax.transAxes,
        fontsize=10,
        verticalalignment='top',
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.6, edgecolor='black', linewidth=1.5))

# Añadir zona de alerta (riesgo de no completar)
ax.axhspan(14, 24, alpha=0.1, color='red', label='Zona de Riesgo')
ax.text(2.5, 18, '⚠️ ZONA DE RIESGO',
        fontsize=11, color='red', fontweight='bold', alpha=0.7,
        ha='center', rotation=0)

# Ajustar layout
plt.tight_layout()

# Guardar el gráfico
output_file = 'sprint2_grafico_burndown.png'
plt.savefig(output_file, dpi=300, bbox_inches='tight', facecolor='white')
print(f"✅ Gráfico guardado en: {output_file}")

# Mostrar el gráfico
plt.show()

# Generar reporte en consola
print("\n" + "="*70)
print("REPORTE DEL SPRINT 2 - GRÁFICO DE BURNDOWN".center(70))
print("="*70)
print(f"\n📅 Período: 6 Octubre - 24 Octubre 2025")
print(f"⏱️  Duración: 3 semanas (15 días hábiles)")
print(f"📊 Total Story Points: 21 SP")
print(f"📍 Estado actual: 14 Octubre 2025 (Semana 2, Día 6)\n")
print(df.to_string(index=False))
print("\n" + "="*70)
print("\n📋 Historias comprometidas:")
print("  • IOC-012: Filtrar datos por línea, periodo y taller (8 SP) - ⏳ En Progreso")
print("  • IOC-008: Comparar desempeño entre turnos y líneas (13 SP) - ⏳ En Progreso")
print("\n✅ Progreso Semana 1 (6-10 Oct):")
print("  • Spike técnico Metabase JWT (SPIKE-01) - ✅ Completado")
print("  • Spike técnico Flying Saucer PDF (SPIKE-02) - ✅ Completado")
print("  • Backend inicial: FilterController + endpoints (BE-TASK-25, 26, 27) - ✅")
print("  • Backend inicial: AnalyticsController + Service (BE-TASK-30, 31, 32) - ✅")
print("  • Backend: FilterService con caché (BE-TASK-27, 28, 29) - ✅")
print("  • Backend: ReportExportService Excel/PDF (BE-TASK-33, 34, 35) - ✅")
print("  • Documentación Swagger y tests básicos (DOC-TASK-01, 02) - ✅")
print("  • Total tareas completadas: 17/17 (Semana 1)")
print("\n⏳ Pendiente para Semanas 2-3:")
print("  • Optimizaciones backend (índices, caché Redis)")
print("  • Tests de integración y E2E")
print("  • Frontend completo (componentes de filtros y comparativas)")
print("  • Integración frontend-backend")
print("  • CI/CD y deployment")
print("\n⚠️  ALERTA:")
print("  • Progreso: 5 SP completados de 21 SP (24%)")
print("  • Esperado a esta altura: ~7 SP (33%)")
print("  • Desviación: -2 SP (9% por debajo del plan)")
print("  • Acción requerida: Acelerar en Semana 2 para recuperar ritmo")
print("="*70)

# Cálculo de métricas adicionales
print("\n📊 MÉTRICAS DEL SPRINT 2 (Actualizado: 14 Oct 2025)")
print("="*70)
print(f"Velocity objetivo: ~7 SP/semana")
print(f"Velocity Semana 1: 5 SP (71% del objetivo)")
print(f"Tareas backend completadas: 17/17 (100% Semana 1)")
print(f"Story Points restantes: 16 SP")
print(f"Semanas restantes: 2 semanas")
print(f"Velocity requerida: 8 SP/semana para completar el sprint")
print(f"\n⚠️  Riesgo: {'ALTO' if 16/2 > 7 else 'MEDIO'} - Se requiere aumentar velocity en 14%")
print("="*70)

# OPCIONAL: Descargar el archivo automáticamente en Colab
print("\n💾 Para descargar el gráfico en Google Colab, ejecuta:")
print("from google.colab import files")
print(f"files.download('{output_file}')")

```
